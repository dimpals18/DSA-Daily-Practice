```python
class Solution {
    public boolean search(int[] nums, int target) {
      Arrays.sort(nums);
      int n =nums.length;
      int start=0;
      int end=n-1;

      while(start<=end){
        int mid = (start+end)/2;
        if(nums[mid]==target){
            return true;
        }else if(nums[mid]<target){
            start=mid+1;
        }else{
           end=mid-1;
        }
      }
      return false;
    }
}
```
