```python
class Solution {
    public int removeElement(int[] nums, int val) {
        int n =nums.length;
        int slow=0;
        int fast=n-1;
       while(slow<=fast){
        if(nums[slow]==val){
           nums[slow]=nums[fast];
           fast--;
        }else{
            slow++;
        }

       }
       return slow;

    }
}
```
