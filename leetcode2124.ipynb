```python
class Solution {
    public boolean checkString(String s) {
        int n= s.length();
        for(int i =1;i<n;i++){
        char prev = s.charAt(i);
        char next = s.charAt(i-1);
        if(prev<next){
            return false;
        }

    }
     return true;
}
}
```
