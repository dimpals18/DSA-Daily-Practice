```python
class Solution {
    public int majorityElement(int[] nums) {
       Arrays.sort(nums);
       int candidate=nums[0];
       int count=1;
       int n =nums.length;
       for(int i =1;i<n;i++){
        if(nums[i]==nums[i-1]){
            count++;
            if(count>n/2){
                return nums[i];
            }
        }else{
            count=1;
            candidate=nums[i];

        }
       }

         return candidate;
    }
}
```
